and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
import os
import json
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
OLLAMA_BASE_URL = 'http://localhost:11434/v1'
```

```python
# set up environment
load_dotenv(override=True)
key = os.getenv("OPENAI_API_KEY")
```

```python
# here is the question; type over this to ask something new

system_prompt = """
Please explain the question asked by the user in an clear and concise manner, and respond in  markdown format.
"""

user_prompt = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming
openai = OpenAI()
stream = openai.chat.completions.create(
    model=MODEL_GPT, 
    messages=[{"role": "system", "content": system_prompt}, {"role": "user", "content": user_prompt}], 
    stream=True
)
response = ""
display_handler = display(Markdown(response), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    update_display(Markdown(response), display_id=display_handler.display_id)
```

```python
# Get Llama 3.2 to answer
ollama_handler = OpenAI(base_url=OLLAMA_BASE_URL, api_key='random')
response = ollama_handler.chat.completions.create(
    model=MODEL_LLAMA, 
    messages=[{"role": "system", "content": system_prompt}, {"role": "user", "content": user_prompt}], 
)
display(Markdown(response.choices[0].message.content))
print('--------------------------------------------')
print(response.choices[0].message.content)
```
